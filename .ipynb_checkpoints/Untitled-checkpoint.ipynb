```python
import pandas as pd

df = pd.read_csv('C:/Users/WINDOWS 11/OneDrive/Документы/NeoMart_Storytelling_Dataset.csv',
                delimiter = ',',
                encoding = 'utf-8')

df['Order_Date'] = pd.to_datetime(df['Order_Date'])

df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month_name()
df['Month_No'] = df['Order_Date'].dt.month
df['Quarter'] = df['Order_Date'].dt.quarter
df['Day'] = df['Order_Date'].dt.day


monthly_sales = (
    df.groupby(['Month_No', 'Month'])['Sales']
      .sum()
      .reset_index()
      .sort_values('Month_No')
)



plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales['Month'],
    monthly_sales['Sales'],
    marker='o',
    linewidth=2,
    linestyle='-'
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout() 
plt.show()                                                      #visualization 1
```

```python
#The monthly sales trend fluctuated through the course of the year. 
#Sales dropped sharply from January to February, recovered between March and May, declined again in June and July, and then increased steadily from August, reaching the highest sales in December.
#This indicates that sales performance was inconsistent, which may be due to seasonal demand, promotional campaigns, changes in customer purchasing behavior, or market conditions. 
#Despite these fluctuations, the business maintained sales activity throughout the year, suggesting opportunities to stabilize revenue through improved sales planning and targeted marketing initiatives.
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.ticker as mticker
import matplotlib.patheffects as pe


NAVY      = "#12294D"   # headers / titles
INK       = "#2B2F38"   # body text
SUBTLE    = "#7A8699"   # secondary text / gridlines
CARD_BG   = "#FFFFFF"
CARD_EDGE = "#E3E7EE"
PAGE_BG   = "#F5F7FA"

SALES_C   = "#2E6F9E"   # brand blue  -> sales metric
PROFIT_C  = "#1E8F6F"   # brand teal  -> profit metric
HILITE    = "#D64550"   # coral       -> annotations / alerts

REGION_COLORS = {"West": "#4C72B0", "North": "#DD8452", "South": "#D64550", "East": "#55A868"}
REGION_ORDER  = ["West", "North", "South", "East"]

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "text.color": INK,
    "axes.labelcolor": INK,
    "xtick.color": INK,
    "ytick.color": INK,
})

rng = np.random.default_rng(7)                                                               #palette


months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
sales  = [161000,132000,152500,175500,176000,167000,141500,175000,190500,196000,185000,201500]
profit = [21400,13200,16700,22800,19500,18900,20000,23400,26300,24800,24900,24600]

regions_by_sales = ["South", "North", "West", "East"]
region_sales = {"South": 735000, "North": 458000, "West": 450000, "East": 405000}

sat_params = {"West": (4.2, 0.35), "North": (4.2, 0.35), "South": (3.2, 0.55), "East": (4.2, 0.35)}
satisfaction = {r: np.clip(rng.normal(m, s, 200), 1.2, 5.0) for r, (m, s) in sat_params.items()}

total_sales = sum(sales)
total_profit = sum(profit)
avg_satisfaction = np.mean([satisfaction[r].mean() for r in REGION_ORDER])
top_region = max(region_sales, key=region_sales.get)

def money(x, pos=None):
    return f"${x/1000:,.0f}K"                                                                    #data


fig = plt.figure(figsize=(14.5, 11.2), facecolor=PAGE_BG)
gs = gridspec.GridSpec(
    3, 4, figure=fig,
    height_ratios=[0.10, 0.40, 0.40],
    hspace=0.55, wspace=0.35,
    left=0.055, right=0.97, top=0.905, bottom=0.115
)


fig.text(0.055, 0.965, "NeoMart Retail Performance", fontsize=23,
          fontweight="bold", color=NAVY, ha="left")
fig.text(0.055, 0.935, "Mini Dashboard  \u2014  Sales, Profitability & Customer Experience",
          fontsize=12.5, color=SUBTLE, ha="left")
fig.text(0.97, 0.95, "FY Overview", fontsize=10.5, color=SUBTLE, ha="right", style="italic")         #header


kpis = [
    ("Total Sales", money(total_sales), SALES_C),
    ("Total Profit", money(total_profit), PROFIT_C),
    ("Top Region", f"{top_region}  ({money(region_sales[top_region])})", REGION_COLORS[top_region]),
    ("Avg. Satisfaction", f"{avg_satisfaction:.1f} / 5.0", NAVY),
]
for i, (label, value, color) in enumerate(kpis):
    axk = fig.add_subplot(gs[0, i])
    axk.axis("off")
    axk.add_patch(plt.Rectangle((0, 0), 1, 1, transform=axk.transAxes,
                                 facecolor=CARD_BG, edgecolor=CARD_EDGE, linewidth=1,
                                 zorder=0, clip_on=False))
    axk.add_patch(plt.Rectangle((0, 0), 0.012, 1, transform=axk.transAxes,
                                 facecolor=color, edgecolor="none", zorder=1, clip_on=False))
    axk.text(0.08, 0.66, label, transform=axk.transAxes, fontsize=10.5,
             color=SUBTLE, ha="left", va="center")
    axk.text(0.08, 0.30, value, transform=axk.transAxes, fontsize=15.5,
             color=NAVY, ha="left", va="center", fontweight="bold")

def style_card(ax):
    ax.set_facecolor(CARD_BG)
    for s in ax.spines.values():
        s.set_visible(True); s.set_color(CARD_EDGE); s.set_linewidth(1.2)
    ax.tick_params(colors=INK, labelsize=9.5)
    ax.grid(axis="y", color="#EDF0F5", linewidth=1, zorder=0)
    ax.set_axisbelow(True)                                                                       #KPI cards 


ax1 = fig.add_subplot(gs[1, 0:2])
style_card(ax1)
ax1.plot(months, sales, marker="o", color=SALES_C, linewidth=2.4, markersize=6,
          markerfacecolor="white", markeredgecolor=SALES_C, markeredgewidth=1.8, zorder=3)
ax1.fill_between(months, sales, min(sales) - 6000, color=SALES_C, alpha=0.08, zorder=1)
ax1.set_title("Monthly Sales Trend", fontsize=13.5, color=NAVY, fontweight="bold", pad=12, loc="left")
ax1.set_ylabel("Total Sales", fontsize=10.5)
ax1.yaxis.set_major_formatter(mticker.FuncFormatter(money))
ax1.tick_params(axis="x", rotation=40)
ax1.set_ylim(min(sales) - 8000, max(sales) + 16000)

peak_i, trough_i = int(np.argmax(sales)), int(np.argmin(sales))
ax1.annotate(f"Peak: {money(sales[peak_i])}", xy=(peak_i, sales[peak_i]),
             xytext=(peak_i - 2.3, sales[peak_i] + 9000),
             fontsize=9.3, color=NAVY, fontweight="bold",
             arrowprops=dict(arrowstyle="-", color=SUBTLE, lw=1))
ax1.annotate(f"Trough: {money(sales[trough_i])}", xy=(trough_i, sales[trough_i]),
             xytext=(trough_i + 0.4, sales[trough_i] - 7500),
             fontsize=9.3, color=HILITE, fontweight="bold",
             arrowprops=dict(arrowstyle="-", color=HILITE, lw=1))                          #monthly sales trend


ax2 = fig.add_subplot(gs[1, 2:4])
style_card(ax2)
ax2.plot(months, profit, marker="o", color=PROFIT_C, linewidth=2.4, markersize=6,
          markerfacecolor="white", markeredgecolor=PROFIT_C, markeredgewidth=1.8, zorder=3)
ax2.fill_between(months, profit, min(profit) - 1200, color=PROFIT_C, alpha=0.08, zorder=1)
ax2.set_title("Monthly Profit Trend", fontsize=13.5, color=NAVY, fontweight="bold", pad=12, loc="left")
ax2.set_ylabel("Total Profit", fontsize=10.5)
ax2.yaxis.set_major_formatter(mticker.FuncFormatter(money))
ax2.tick_params(axis="x", rotation=40)
ax2.set_ylim(min(profit) - 1800, max(profit) + 3500)

peak_i, trough_i = int(np.argmax(profit)), int(np.argmin(profit))
ax2.annotate(f"Peak: {money(profit[peak_i])}", xy=(peak_i, profit[peak_i]),
             xytext=(peak_i - 2.6, profit[peak_i] + 1600),
             fontsize=9.3, color=NAVY, fontweight="bold",
             arrowprops=dict(arrowstyle="-", color=SUBTLE, lw=1))
ax2.annotate(f"Trough: {money(profit[trough_i])}", xy=(trough_i, profit[trough_i]),
             xytext=(trough_i + 0.7, profit[trough_i] - 2100),
             fontsize=9.3, color=HILITE, fontweight="bold",
             arrowprops=dict(arrowstyle="-", color=HILITE, lw=1))                           #monthly profit trend


ax3 = fig.add_subplot(gs[2, 0:2])
style_card(ax3)
bar_colors = [REGION_COLORS[r] for r in regions_by_sales]
bars = ax3.bar(regions_by_sales, [region_sales[r] for r in regions_by_sales],
                color=bar_colors, width=0.58, zorder=3,
                edgecolor="white", linewidth=0.6)
ax3.set_title("Total Sales by Region", fontsize=13.5, color=NAVY, fontweight="bold", pad=12, loc="left")
ax3.set_ylabel("Total Sales", fontsize=10.5)
ax3.yaxis.set_major_formatter(mticker.FuncFormatter(money))
ax3.set_ylim(0, max(region_sales.values()) * 1.18)
for b, r in zip(bars, regions_by_sales):
    v = region_sales[r]
    ax3.text(b.get_x() + b.get_width()/2, v + 14000, money(v),
              ha="center", fontsize=10, color=NAVY, fontweight="bold")

gap_pct = (region_sales["South"] - region_sales["North"]) / region_sales["North"] * 100
ax3.annotate(f"South outsells the\nnext region by {gap_pct:.0f}%",
             xy=(0, region_sales["South"]), xytext=(0.55, region_sales["South"] * 0.85),
             fontsize=9.3, color=HILITE, fontweight="bold", ha="left",
             arrowprops=dict(arrowstyle="->", color=HILITE, lw=1.3,
                              connectionstyle="arc3,rad=-0.25"))                              #regional sales


ax4 = fig.add_subplot(gs[2, 2:4])
style_card(ax4)
data4 = [satisfaction[r] for r in REGION_ORDER]
bp4 = ax4.boxplot(data4, tick_labels=REGION_ORDER, patch_artist=True, widths=0.5,
                    zorder=3, medianprops=dict(color=NAVY, linewidth=2),
                    boxprops=dict(linewidth=1), whiskerprops=dict(color=SUBTLE),
                    capprops=dict(color=SUBTLE),
                    flierprops=dict(marker="o", markersize=4, markerfacecolor="white",
                                     markeredgecolor=SUBTLE))
for patch, r in zip(bp4["boxes"], REGION_ORDER):
    patch.set_facecolor(REGION_COLORS[r]); patch.set_alpha(0.82); patch.set_edgecolor(NAVY)
ax4.set_title("Customer Satisfaction by Region", fontsize=13.5, color=NAVY, fontweight="bold", pad=12, loc="left")
ax4.set_ylabel("Satisfaction Score (1\u20135)", fontsize=10.5)
ax4.set_ylim(1, 5.6)

south_median = np.median(satisfaction["South"])
ax4.annotate("Lowest & most variable\nsatisfaction",
             xy=(REGION_ORDER.index("South") + 1, south_median),
             xytext=(REGION_ORDER.index("South") + 1.55, 2.0),
             fontsize=9.3, color=HILITE, fontweight="bold", ha="left",
             arrowprops=dict(arrowstyle="->", color=HILITE, lw=1.3,
                              connectionstyle="arc3,rad=0.2"))                          #customer satisfaction


handles = [plt.Rectangle((0, 0), 1, 1, facecolor=REGION_COLORS[r], edgecolor="none") for r in REGION_ORDER]
fig.legend(handles, REGION_ORDER, loc="lower center", ncol=4, frameon=False,
           bbox_to_anchor=(0.5, 0.048), fontsize=10, title="Region",
           title_fontsize=10, labelcolor=INK)

fig.text(0.055, 0.015,
          "Note: monthly and regional totals reflect the reviewed source charts; the satisfaction panel "
          "uses illustrative points calibrated to the reported regional medians.",
          fontsize=8, color=SUBTLE, ha="left")                                       #tie charts together

plt.savefig("neomart_dashboard", dpi=160,
            facecolor=PAGE_BG, bbox_inches="tight")
print("saved")

```
